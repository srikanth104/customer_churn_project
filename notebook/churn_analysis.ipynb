```python

X = df.drop(columns=["Churn"])
y = df["Churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["str"]).columns.tolist()

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier

model_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(random_state=42))
    ]
)

print("Pipeline created successfully.")
```

```python
model_pipeline.fit(X_train, y_train)

print("Model training completed.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred = model_pipeline.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, pos_label="Yes")
recall = recall_score(y_test, y_pred, pos_label="Yes")
f1 = f1_score(y_test, y_pred, pos_label="Yes")

print("Configuration 1")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["No", "Yes"]
)

plt.title("Confusion Matrix - Decision Tree Configuration 1")
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier

model_pipeline_2 = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(
            max_depth=5,
            random_state=42
        ))
    ]
)

model_pipeline_2.fit(X_train, y_train)

print("Configuration 2 training completed.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


y_pred_2 = model_pipeline_2.predict(X_test)

accuracy_2 = accuracy_score(y_test, y_pred_2)
precision_2 = precision_score(y_test, y_pred_2, pos_label="Yes")
recall_2 = recall_score(y_test, y_pred_2, pos_label="Yes")
f1_2 = f1_score(y_test, y_pred_2, pos_label="Yes")

print("Configuration 2")
print("Accuracy :", accuracy_2)
print("Precision:", precision_2)
print("Recall   :", recall_2)
print("F1-score :", f1_2)
```

```python
print("Configuration 2 model exists:", model_pipeline_2 is not None)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

# Display the confusion matrix for Configuration 2.
# This shows how many customers were correctly and incorrectly classified as churn or non-churn.

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_2,
    display_labels=["No", "Yes"]
)

plt.title("Confusion Matrix - Decision Tree Configuration 2")
plt.show()
```

```python
feature_names = model_pipeline_2["preprocessor"].get_feature_names_out()

print("Number of transformed features:", len(feature_names))
print(feature_names)
```

```python
importances = model_pipeline_2["classifier"].feature_importances_

feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance_df.head(10))
```

```python
top_features = feature_importance_df.head(10).sort_values(
    by="Importance"
)

plt.figure(figsize=(9, 6))
plt.barh(top_features["Feature"], top_features["Importance"])

plt.title("Top 10 Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(24, 12))

plot_tree(
    model_pipeline_2["classifier"],
    feature_names=feature_names,
    class_names=["No", "Yes"],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title("Decision Tree - Configuration 2")
plt.show()
```
